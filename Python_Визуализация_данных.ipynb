**Построение графика**


Для начала построим простую линейную зависимость, зададим название графику, подпишем оси и отобразим сетку:


```python
import numpy as np
import matplotlib.pyplot as plt

# Независимая (x) и зависимая (y) переменные
x = np.linspace(0, 30, 90)
y = x  # исправлено имя переменной на 'y'

# Построение графика
plt.title('Линейная зависимость y = x')  # заголовок
plt.xlabel('x')  # ось абсцисс
plt.ylabel('y')  # ось ординат
plt.grid()  # включение отображения сетки
plt.plot(x, y)  # построение графика
plt.show()  # добавлено для отображения графика
```

Изменим тип линии и её цвет, для этого в функцию plot(), в качестве третьего параметра, передадим строку, сформированную определенным образом, в нашем случае это 'г--', где 'г' означает красный цвет, а '--' - тип линии — пунктирная линия:

```python
import numpy as np
import matplotlib.pyplot as plt

# Линейная зависимость
x = np.linspace(0, 10, 50)
y1 = 3 * x + 2  # Линейная функция

# Синусоидальная зависимость
y2 = 5 * np.sin(x)  # Синусоидальная функция

# Построение графика
plt.figure(figsize=(10, 6))  # Установка размера графика
plt.title('Зависимости: y1 = 3x + 2, y2 = 5sin(x)')  # заголовок
plt.xlabel('x')  # ось абсцисс
plt.ylabel('y1, y2')  # ось ординат
plt.grid()  # включение отображения сетки
plt.plot(x, y1, label='y1 = 3x + 2')  # построение графика для линейной зависимости
plt.plot(x, y2, label='y2 = 5sin(x)')  # построение графика для синусоидальной зависимости
plt.legend()  # отображение легенды
plt.show()  # показать график
```

**Представление графиков на разных полях**

ретья, довольно часто встречающаяся задача — это отображение двух или более различных полей, на которых может быть представлено несколько графиков.
Построим уже известные нам зависимости на разных полях:

```python
x= np.linspace(0, 10, 50)

y1=x # Линейная зависимость

y2= [i**2 for i in x] # Квадратичная зависимость
# построение графиков
plt.figure(figsize=(9,9))

plt.subplot(2, 1, 1)

plt.plot(x, y1) # построение графика

plt.title('Зависимости: y1=x, y2= x^2') # заголовок

plt.ylabel('y1', fontsize=14) # ось ординат

plt.grid(True)# включение отображение сетки

plt.subplot(2, 1, 2)

plt.plot(x, y2)# построение графика

plt.xlabel('x', fontsize=14)# ось абсцисс

plt.ylabel('y2', fontsize=14)# ось ординат

plt.grid(True)# включение отображение сетки
```

Здесь мы воспользовались двумя новыми функциями:
•	 figure() — функция для задания глобальных параметров отображения графиков. В неё через параметр figsize передаётся кортеж из двух элементов, определяющий общий размер фигуры.
•	 subplot() — функция для задания местоположения поля с графиком. Существует несколько способов задания областей для вывода графиков. В примере мы воспользовались вариантом, который предполагает передачу трёх аргументов: первый аргумент — количество строк, второй — столбцов в формируемом поле, третий — индекс (номер поля, считаем сверху вниз, слева направо).
Мы использовали параметр fontsize функций xlabel() и ylabel() для задания размера шрифта


**Построение диаграммы для категориальных данных**


До этого мы строили графики по численным данным, то есть зависимая
и независимая переменные имели числовой тип. На практике довольно
часто приходится работать с данными не числовой природы — имена
людей, названия городов и т.п. Построим диаграмму, на которой будет
отображаться количество фруктов в магазине:


```python
import matplotlib.pyplot as plt  # Импортируем библиотеку для построения графиков

# Новые данные - массивы с названиями фруктов и их количествами
fruits = ['strawberry', 'blueberry', 'mango', 'kiwi', 'grape']  # Список названий фруктов
counts = [50, 30, 20, 35, 45]  # Соответствующие количества для каждого фрукта

# Построение столбчатой диаграммы
plt.bar(fruits, counts)  # Создаем столбчатую диаграмму на основе данных о фруктах и их количестве

# Добавляем заголовок и метки к осям
plt.title('Fruits Count')  # Заголовок графика
plt.xlabel('Fruit')  # Подпись оси x (названия фруктов)
plt.ylabel('Count')  # Подпись оси y (количество)

# Показ диаграммы
plt.show()  # Отображаем график на экране
```

Для вывода диаграммы мы использовали функцию bar().

**Основные элементы графика**


Рассмотрим основные термины и понятия, касающиеся изображения графика, с которыми вам необходимо будет познакомиться для того чтобы в дальнейшем не было трудностей при изучении глав этой книги и документации по библиотеке Matplotlib.

Корневым элементом, на котором Matplotlib строит изображение, является фигура (Figure). Всё, что перечислено на рисунке— это элементы фигуры. Рассмотрим её составляющие более подробно.

**График**

На рисунке представлены два графика — линейный и точечный. Matplotlib предоставляет огромное количество различных настроек, которые можно использовать для того, чтобы придать графику требуемый вид: цвет, толщина, тип, стиль линии и многое другое.

**Оси**

Вторым по важности элементом фигуры являются оси. Для каждой оси можно задать метку (подпись), основные (major) и дополнительные (minor) тики, их подписи, размер, толщину и диапазоны.


Сетка и легенда
Сетка и легенда являются элементами фигуры, которые значительно повышают информативность графика. Сетка может быть основной (major) и дополнительной (minor). Каждому типу сетки можно задавать цвет, толщину линии и тип. Для отображения сетки и легенды используются соответствующие команды.
Ниже представлен код, с помощью которого был построен график, изображённый на рисунке


image.png

```python
import matplotlib.pyplot as plt  # Импортируем библиотеку для построения графиков
from matplotlib.ticker import (MultipleLocator, FormatStrFormatter, AutoMinorLocator)  # Импортируем классы для настройки делений на осях
import numpy as np  # Импортируем библиотеку для работы с числовыми массивами

# Создаем массив значений x от 0 до 10 с 10 точками
x = np.linspace(0, 10, 10)
y1 = 4*x  # Вычисляем y1 как 4x
y2 = [i**2 for i in x]  # Вычисляем y2 как x^2 с помощью списка

# Создаем фигуру и ось для графика
fig, ax = plt.subplots(figsize=(8, 6))  # Размер графика: 8 на 6 дюймов

# Добавляем заголовок и метки осей
ax.set_title('Графики зависимостей: y1=4*x, y2=x^2', fontsize=16)  # Заголовок графика
ax.set_xlabel('x', fontsize=14)  # Подпись оси x
ax.set_ylabel('y1, y2', fontsize=14)  # Подпись оси y

# Настраиваем сетку на графике
ax.grid(which='major', linewidth=1.2)  # Основная сетка
ax.grid(which='minor', linestyle='--', color='gray', linewidth=0.5)  # Второстепенная сетка

# Добавляем точки y1 на график
ax.scatter(x, y1, c='red', label='y1 = 4*x')  # Отображение точек y1 красным цветом
# Добавляем линию y2 на график
ax.plot(x, y2, label='y2 = x^2')  # Отображение y2 линией

ax.legend()  # Показываем легенду, чтобы обозначить, какая линия за что отвечает

# Настраиваем деления на осях
ax.xaxis.set_minor_locator(AutoMinorLocator())  # Автоматическое добавление второстепенных делений по оси x
ax.yaxis.set_minor_locator(AutoMinorLocator())  # Автоматическое добавление второстепенных делений по оси y

# Настраиваем параметры отображения делений
ax.tick_params(which='major', length=10, width=2)  # Основные деления: длина 10, ширина 2
ax.tick_params(which='minor', length=5, width=1)  # Второстепенные деления: длина 5, ширина 1
```
